```python
import re
metals = ['Li', 'Na', 'K', 'Rb', 'Cs', 'Fr', 'Be', 'Mg', 'Ca', 'Sr', 'Ba', 'Ra', 'Sc', 'Ti', 
          'V', 'Cr', 'Mn', 'Fe', 'Co', 'Ni', 'Cu', 'Zn', 'Y', 'Zr', 'Nb', 'Mo', 'Tc', 'Ru', 
          'Rh', 'Pd', 'Ag', 'Cd', 'Hf', 'Ta', 'W', 'Re', 'Os', 'Ir', 'Pt', 'Au', 'Hg', 'La', 
          'Ce', 'Pr', 'Nd', 'Pm', 'Sm', 'Eu', 'Gd', 'Tb', 'Dy', 'Ho', 'Er', 'Tm', 'Yb', 'Lu', 
          'Th', 'Pa', 'U', 'Np', 'Pu', 'Am', 'Cm', 'Bk', 'Cf', 'Es', 'Fm', 'Md', 'No', 'Lr', 
          'Al', 'Ga', 'In', 'Tl', 'Pb', 'Bi', 'Po']

# Create a regex pattern that matches any of the metal elements
metal_pattern = re.compile(r'\b(' + '|'.join(metals) + r')\b')

# Create a regex pattern to detect organic elements (C, H, O, N, P, S)
organic_pattern = re.compile(r'[CHONPS]')


with open('pubchem-10m.txt', 'r') as file:
    smiles_data = file.readlines()

```

```python

filtered_smiles = []
removed_smiles = []

# Filtering SMILES
for smiles in smiles_data:
    smiles_str = smiles.strip()
    
    # Check if it contains any metal
    if metal_pattern.search(smiles_str):
        # If it also contains organic elements, keep it
        if organic_pattern.search(smiles_str):
            filtered_smiles.append(smiles_str)  # Keep it
        else:
            removed_smiles.append(smiles_str)  # Remove it
    else:
        filtered_smiles.append(smiles_str)  # Keep it since no metal present
```
